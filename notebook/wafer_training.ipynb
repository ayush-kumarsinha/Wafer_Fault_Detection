FEATURE ENGINEERING

```python
df_fe = df.copy()
```

```python
df_fe = df_fe.drop(columns=['Unnamed: 0'])
```

```python
df_fe.shape
```

```python
X = df_fe.drop('Good/Bad', axis=1)

y = df_fe['Good/Bad']
```

```python
(X.isnull().sum().sum())
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='median')

X = pd.DataFrame(
    imputer.fit_transform(X),
    columns=X.columns
)
```

```python
from sklearn.feature_selection import VarianceThreshold

selector = VarianceThreshold(threshold=0)

X = pd.DataFrame(
    selector.fit_transform(X)
)

print(X.shape)
```

```python
missing_percent = (df_fe.isnull().sum()/len(df_fe))*100

high_missing_cols = missing_percent[
    missing_percent > 50
].index

print("Columns to remove:", len(high_missing_cols))
```

TRAIN/TEST SPLIT

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```python
!pip install imbalanced-learn
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train,
    y_train
)

print(y_train_smote.value_counts())
```

```python
print(X.shape)
print(X_train_smote.shape)
print(y_train_smote.value_counts())
```

MODEL TRAINING


APPLY RANDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train_smote, y_train_smote)

rf_pred = rf.predict(X_test)

print(classification_report(y_test, rf_pred))
```

```python
y_train_smote = y_train_smote.replace({
    -1: 0,
     1: 1
})

y_test_xgb = y_test.replace({
    -1: 0,
     1: 1
})
```

```python
print(y_train_smote.value_counts())
print(y_test_xgb.value_counts())
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    random_state=42,
    eval_metric='logloss'
)

xgb.fit(X_train_smote, y_train_smote)

xgb_pred = xgb.predict(X_test)

print(classification_report(y_test, xgb_pred))
```

```python
from sklearn.metrics import accuracy_score, f1_score

models = {
    "Random Forest": rf_pred,
    "XGBoost": xgb_pred
}

for name, pred in models.items():
    print(f"\n{name}")
    print("Accuracy:", accuracy_score(y_test, pred))
    print("F1 Score:", f1_score(y_test, pred))
```
